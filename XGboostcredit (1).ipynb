```python
 import pandas as pd

# Assuming the uploaded file is 'credit ecotrix dataset.csv'
df = pd.read_csv('credit ecotrix dataset.csv')

# Display the first 5 rows of the DataFrame
display(df.head(10))
display(df.describe())
```

```python
# using hot hand encoder
categorical= ['SEX', 'EDUCATION','MARRIAGE','PAY_0','PAY_2','PAY_3','PAY_4','PAY_5','PAY_6']
non_categorical = ['ID', 'LIMIT_BAL', 'AGE', 'BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6', 'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']
```

```python
cols_to_encode = [col for col in categorical if col in df.columns]

if cols_to_encode:
    df = pd.get_dummies(df, columns=cols_to_encode)
    print(f"Successfully created dummy variables for: {cols_to_encode}")
else:
    print("Categorical columns already processed or not found in DataFrame. No new dummy variables created.")

df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
test_df = df.sample(frac=0.2, random_state=42)
train_df = df.drop(test_df.index)
train_df.shape, test_df.shape
```

```python
means = train_df[non_categorical].mean()
sd = train_df[non_categorical].std()
train_df[non_categorical] = (train_df[non_categorical] - means) / sd
test_df[non_categorical] = (test_df[non_categorical] - means) / sd
train_df.head()
```

```python
from sklearn.model_selection import train_test_split
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Define the target variable
TARGET = 'default.payment.next.month'

# Separate features (X) and target (y) for training and testing sets
# Drop the 'ID' column from features as it's not relevant for prediction
X_train = train_df.drop([TARGET, 'ID'], axis=1)
y_train = train_df[TARGET]
X_test = test_df.drop([TARGET, 'ID'], axis=1)
y_test = test_df[TARGET]

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

```python
# Initialize the XGBoost Classifier
model = xgb.XGBClassifier(
    objective='binary:logistic', # For binary classification
    eval_metric='logloss',       # Evaluation metric
    use_label_encoder=False,     # Suppress warning for deprecated parameter
    random_state=42
)

# Train the model
model.fit(X_train, y_train)

print("XGBoost Model training complete.")
```

```python
# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
display(pd.DataFrame(confusion_matrix(y_test, y_pred), index=['Actual No Default', 'Actual Default'], columns=['Predicted No Default', 'Predicted Default']))
```

### Model Performance Visualizations

To better understand the model's performance, especially given the class imbalance, let's visualize the Confusion Matrix, ROC Curve, and Precision-Recall Curve.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_curve, auc, precision_recall_curve

# 1. Confusion Matrix Heatmap
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds',
            xticklabels=['Predicted No Default', 'Predicted Default'],
            yticklabels=['Actual No Default', 'Actual Default'])
plt.title('Confusion Matrix Heatmap')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

```python
# ROC curve, Get predicted probabilities for the positive(default = 1)
y_pred_proba = model.predict_proba(X_test)[:, 1]

fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.show()
```

```python
# 3. Precision-Recall Curve
precision, recall, _ = precision_recall_curve(y_test, y_pred_proba)

plt.figure(figsize=(8, 6))
plt.plot(recall, precision, color='blue', lw=2, label='Precision-Recall curve')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.ylim([0.0, 1.05])
plt.xlim([0.0, 1.0])
plt.show()
```

# #solving the inbalance in dataset problem and adjusting the threshold

```python
#giving more weight to the minority class (defaults) to rebalancce our dataset
count_negative_samples = y_train.value_counts()[0] # Assuming 0 is the non-default class
count_positive_samples = y_train.value_counts()[1] # Assuming 1 is the default class

scale_pos_weight_value = count_negative_samples / count_positive_samples
print(f"Calculated scale_pos_weight: {scale_pos_weight_value:.2f}")
```

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Initialize the XGBoost Classifier with scale_pos_weight
model_balanced = xgb.XGBClassifier(
    objective='binary:logistic',
    eval_metric='logloss',
    use_label_encoder=False, # Suppress warning for deprecated parameter
    random_state=42,
    scale_pos_weight=scale_pos_weight_value # Apply the calculated weight
)

# Train the balanced model
model_balanced.fit(X_train, y_train)

print("XGBoost Model (with balanced weights) training complete.")

# Make predictions on the test set with the balanced model
y_pred_balanced = model_balanced.predict(X_test)

# Evaluate the balanced model
accuracy_balanced = accuracy_score(y_test, y_pred_balanced)
print(f"\nAccuracy (Balanced Model): {accuracy_balanced:.4f}")

print("\nClassification Report (Balanced Model):")
print(classification_report(y_test, y_pred_balanced))

print("\nConfusion Matrix (Balanced Model):")
display(pd.DataFrame(confusion_matrix(y_test, y_pred_balanced), index=['Actual No Default', 'Actual Default'], columns=['Predicted No Default', 'Predicted Default']))
```

You should observe that while the overall accuracy might slightly decrease, the **recall for class 1 (defaulters) has likely increased**. This means the model is now better at identifying actual defaulters, which is often more critical in scenarios like credit risk assessment than overall accuracy. There's usually a trade-off between precision and recall when addressing class imbalance.

We can also visualize the performance of this balanced model, similar to what we did before, to see the changes in the confusion matrix and PR curve more clearly.

#  the Classification Threshold

### Optimal Threshold Criteria

For our case, since we want to improve the **detection of defaulters** (which means increasing recall for the positive class) while still maintaining a reasonable balance with precision, optimizing the **F1-score** is a good starting point. Let's visualize the precision, recall, and F1-score across different thresholds and find the one that maximizes F1-score for our `model_balanced`.

```python
from sklearn.metrics import precision_recall_curve, f1_score

# Get predicted probabilities from the balanced model for the positive class
y_pred_proba_balanced = model_balanced.predict_proba(X_test)[:, 1]

# Calculate precision, recall, and thresholds
precisions, recalls, thresholds = precision_recall_curve(y_test, y_pred_proba_balanced)

# Calculate F1-scores for each threshold
f1_scores = [f1_score(y_test, (y_pred_proba_balanced >= t)) for t in thresholds]

# Find the threshold that maximizes the F1-score
optimal_threshold_idx = np.argmax(f1_scores)
optimal_threshold = thresholds[optimal_threshold_idx]

print(f"Optimal Threshold (maximizing F1-score): {optimal_threshold:.4f}")
print(f"Precision at optimal threshold: {precisions[optimal_threshold_idx]:.4f}")
print(f"Recall at optimal threshold: {recalls[optimal_threshold_idx]:.4f}")
print(f"F1-score at optimal threshold: {f1_scores[optimal_threshold_idx]:.4f}")

# Plot Precision, Recall, and F1-score vs. Threshold
plt.figure(figsize=(10, 6))
plt.plot(thresholds, precisions[:-1], "b--", label="Precision")
plt.plot(thresholds, recalls[:-1], "g-", label="Recall")
plt.plot(thresholds, f1_scores, "r:", label="F1-score")
plt.vlines(optimal_threshold, 0, 1, color='gray', linestyle='--', label=f'Optimal Threshold ({optimal_threshold:.2f})')
plt.xlabel("Threshold")
plt.ylabel("Score")
plt.title("Precision, Recall, and F1-score vs. Threshold")
plt.legend()
plt.grid(True)
plt.show()
```

Now, let's re-evaluate the balanced model's performance using this `optimal_threshold` and compare the classification report and confusion matrix to see the impact.

```python
from sklearn.metrics import classification_report, confusion_matrix

# Make predictions using the optimal threshold
y_pred_optimal_threshold = (y_pred_proba_balanced >= optimal_threshold).astype(int)

print(f"\nClassification Report (Balanced Model with Optimal Threshold {optimal_threshold:.4f}):")
print(classification_report(y_test, y_pred_optimal_threshold))

print("\nConfusion Matrix (Balanced Model with Optimal Threshold):")
display(pd.DataFrame(confusion_matrix(y_test, y_pred_optimal_threshold), index=['Actual No Default', 'Actual Default'], columns=['Predicted No Default', 'Predicted Default']))
```

By adjusting the threshold, you should notice a further improvement in the recall for the defaulter class (class 1), often with a corresponding change in precision and potentially overall accuracy. This highlights the importance of choosing a threshold based on your specific problem's needs rather than just sticking to the default 0.5.

## Model Explanation using SHAP

To understand how each feature contributes to the model's predictions, we will use SHAP (SHapley Additive exPlanations). SHAP values help us interpret the output of machine learning models by explaining the prediction of an instance by computing the contribution of each feature to the prediction.

```python
import shap

# Initialize Javascript for SHAP visualizations
shap.initjs()

# For computational efficiency, let's sample a subset of the test data for explanation
# Using the X_test from the original split, before any oversampling or weighting effects on training
X_test_sample = X_test.sample(n=1000, random_state=42)

# Create a SHAP explainer object for our balanced XGBoost model
explainer = shap.TreeExplainer(model_balanced)

# Calculate SHAP values for the sampled test data
# Note: This can take some time depending on the sample size and model complexity
shap_values = explainer.shap_values(X_test_sample)

print("SHAP values calculated.")
```

### SHAP Summary Plot

The summary plot combines feature importance with feature effects. Each point on the plot is a shapley value for that feature and that instance. The position on the y-axis is determined by the feature, and the x-axis position by the shap value. The color of the point indicates the feature's value (red = high, blue = low). Overlapping points are stacked to show density.

```python
# Visualize the SHAP values using a summary plot
# This plot shows the most important features and their impact (positive or negative) on the model output
shap.summary_plot(shap_values, X_test_sample, plot_type="dot", max_display=20)
```


 SHAP plot confirms that the model’s decisions are primarily driven by recent payment behavior and credit limits. High payment delays sharply increase default risk, while higher limits and payments reduce it. However, the model may rely too heavily on short‑term payment features, so improving feature diversity and reducing redundancy could make it more robust.


### SHAP Dependence Plots

Dependence plots show the effect of a single feature across the whole dataset. They can reveal if the relationship between a feature and the predicted outcome is linear, monotonic, or more complex. We can also add an interaction feature to see how two features jointly affect the prediction.

```python
# Example dependence plot for a top feature (e.g., 'LIMIT_BAL' or 'AGE' as seen from the summary plot)
# You can choose any feature from the summary plot to examine its relationship more closely.

# For instance, let's pick 'LIMIT_BAL' as it often shows high importance in credit datasets.
# If 'LIMIT_BAL' is not in the top features, choose another prominent one from the summary plot.
feature_to_plot = 'LIMIT_BAL'

# Check if the feature exists in the sampled data, otherwise pick another
if feature_to_plot not in X_test_sample.columns:
    print(f"Feature '{feature_to_plot}' not found. Please choose an existing feature from X_test_sample.")
else:
    plt.figure(figsize=(8, 6))
    shap.dependence_plot(
        feature_to_plot,
        shap_values,
        X_test_sample,
        interaction_index=None, # You can specify an interaction_index to see interaction effects
        title=f"SHAP Dependence Plot for {feature_to_plot}"
    )

# Example with an interaction feature (e.g., 'AGE')
# This helps to see how the effect of `feature_to_plot` changes depending on the value of 'AGE'.
interaction_feature = 'AGE'

if feature_to_plot in X_test_sample.columns and interaction_feature in X_test_sample.columns:
    plt.figure(figsize=(10, 7))
    shap.dependence_plot(
        feature_to_plot,
        shap_values,
        X_test_sample,
        interaction_index=interaction_feature,
        title=f"SHAP Dependence Plot for {feature_to_plot} with {interaction_feature} interaction"
    )
```

SHAP Summary Plot: Understanding Overall Feature Impact
What it shows: This plot gives us a global view of feature importance and how each feature influences the model's output (the probability of default). Each row represents a feature, ordered by its overall importance. Each point is a single data instance from your test set. The horizontal position of the point indicates the SHAP value, showing whether that feature's value pushes the prediction higher (towards default, to the right) or lower (away from default, to the left).

Color interpretation: The color of each point tells us the original value of the feature for that instance. Red typically means a high feature value, and blue means a low feature value.

How to interpret for credit default:

Identify Key Drivers: Look at the features at the top of the plot. These are the most influential in determining default risk. For example, if 'PAY_0' (repayment status in September) is high on the list, it means it's a very strong predictor.
Direction of Impact: For each feature, observe the spread of red and blue points. If many red points (high values) are on the right side of the plot (positive SHAP values), it means that high values of that feature increase the probability of default. Conversely, if blue points (low values) are on the left, low values of that feature decrease the probability of default. For instance, you might see:
LIMIT_BAL: Often, higher LIMIT_BAL (red dots) might have negative SHAP values (pulling to the left), indicating that higher credit limits are associated with a lower probability of default.
PAY_X features (e.g., PAY_0, PAY_2): High values here (indicating delayed payments, shown in red) would likely have positive SHAP values (pushing to the right), meaning severe payment delays strongly increase the probability of default.
AGE: Its impact might be more nuanced, with certain age ranges (e.g., very young or very old) potentially having a higher SHAP value in one direction or another.
This plot helps you grasp, at a glance, which factors are most critical and in what general direction they push the model's prediction for thousands of different customers.

SHAP Dependence Plots: Unpacking Individual Feature Relationships
What it shows: These plots zoom in on a single feature to show how its value precisely affects the model's prediction. The x-axis represents the actual values of the chosen feature (e.g., LIMIT_BAL), and the y-axis shows the SHAP value of that feature. Each point represents an individual data point from your sample.

Interaction Feature (color): When an interaction feature is specified (like AGE in your second dependence plot), the points are colored according to the value of that interaction feature. This reveals how the impact of the primary feature changes depending on the value of the secondary feature.

How to interpret for credit default:

Feature LIMIT_BAL (without interaction): This plot will show how changes in the credit limit directly influence the model's output. You can see if the relationship is linear (e.g., a constant decrease in default probability as LIMIT_BAL increases) or non-linear (e.g., the effect might flatten out after a certain credit limit). This helps confirm intuitive relationships or reveal unexpected ones.
Feature LIMIT_BAL with AGE interaction: This is particularly insightful. The plot will show LIMIT_BAL on the x-axis and its SHAP value on the y-axis, but the points will be colored by AGE. For instance, you might observe that for customers with a lower AGE (blue points), a given LIMIT_BAL might have a more significant negative SHAP value (meaning it strongly reduces default probability) compared to older customers (red points) with the same LIMIT_BAL. This indicates that the effect of a credit limit on default risk is moderated by the customer's age, uncovering a complex interaction within the model.
