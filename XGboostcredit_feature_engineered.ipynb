### Feature Engineering: Aggregating Bill/Payment Series and Creating Ratios

To capture more comprehensive information from the monthly bill and payment statements, we will create new features by:

1.  **Aggregating `BILL_AMT` and `PAY_AMT`:** Calculating the sum and mean of these amounts over the six-month period.
2.  **Creating Utilization Ratios:** Dividing aggregated bill amounts by the `LIMIT_BAL` to understand how much of the credit limit is being used.
3.  **Creating Payment Ratios:** Dividing aggregated payment amounts by aggregated bill amounts to assess repayment behavior relative to outstanding bills.

```python
# Define the columns for bill amounts and payment amounts
bill_amt_cols = [f'BILL_AMT{i}' for i in range(1, 7)]
pay_amt_cols = [f'PAY_AMT{i}' for i in range(1, 7)]

def create_engineered_features(df):
    # Aggregate BILL_AMT columns
    df['BILL_AMT_sum'] = df[bill_amt_cols].sum(axis=1)
    df['BILL_AMT_mean'] = df[bill_amt_cols].mean(axis=1)

    # Aggregate PAY_AMT columns
    df['PAY_AMT_sum'] = df[pay_amt_cols].sum(axis=1)
    df['PAY_AMT_mean'] = df[pay_amt_cols].mean(axis=1)

    # Create Utilization Ratios
    # Handle cases where LIMIT_BAL might be zero to avoid division by zero
    df['UTIL_RATIO_sum'] = df.apply(lambda row: row['BILL_AMT_sum'] / row['LIMIT_BAL'] if row['LIMIT_BAL'] != 0 else 0, axis=1)
    df['UTIL_RATIO_mean'] = df.apply(lambda row: row['BILL_AMT_mean'] / row['LIMIT_BAL'] if row['LIMIT_BAL'] != 0 else 0, axis=1)

    # Create Payment Ratios
    # Handle cases where BILL_AMT_sum or BILL_AMT_mean might be zero
    df['PAY_TO_BILL_RATIO_sum'] = df.apply(lambda row: row['PAY_AMT_sum'] / row['BILL_AMT_sum'] if row['BILL_AMT_sum'] != 0 else 0, axis=1)
    df['PAY_TO_BILL_RATIO_mean'] = df.apply(lambda row: row['PAY_AMT_mean'] / row['BILL_AMT_mean'] if row['BILL_AMT_mean'] != 0 else 0, axis=1)

    return df

# Apply feature engineering to both training and test dataframes
train_df_fe = create_engineered_features(train_df.copy())
test_df_fe = create_engineered_features(test_df.copy())

print("New features added to train_df_fe and test_df_fe.")
print("Shape of train_df_fe:", train_df_fe.shape)
print("Shape of test_df_fe:", test_df_fe.shape)

display(train_df_fe.head())
```

### Correlation Analysis and Feature Selection

Highly correlated features can introduce redundancy and multicollinearity, potentially affecting model stability and interpretability. We will compute the correlation matrix and remove one of the features from any pair that has a correlation coefficient above a certain threshold (e.g., 0.95).

```python
# Calculate the correlation matrix for the training data
correlation_matrix = train_df_fe.corr().abs()

# Select upper triangle of correlation matrix
upper_tri = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))

# Find features with correlation greater than 0.95
to_drop = [column for column in upper_tri.columns if any(upper_tri[column] > 0.95)]

print(f"Features to drop due to high correlation: {to_drop}")

# Drop features from both training and test dataframes
train_df_cleaned = train_df_fe.drop(columns=to_drop)
test_df_cleaned = test_df_fe.drop(columns=to_drop)

print("Shape of train_df_cleaned after dropping highly correlated features:", train_df_cleaned.shape)
print("Shape of test_df_cleaned after dropping highly correlated features:", test_df_cleaned.shape)
```

### Retraining XGBoost with New Features and Regularization

Now, with the new engineered features and after handling highly correlated columns, we will retrain the XGBoost model. We will re-use the `scale_pos_weight` for class imbalance and introduce additional regularization parameters to prevent overfitting.

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Define the target variable (assuming 'default.payment.next.month' is still the target)
TARGET = 'default.payment.next.month'

# Re-separate features (X) and target (y) for training and testing sets
# Drop the 'ID' column and the TARGET variable from features
X_train_new = train_df_cleaned.drop([TARGET, 'ID'], axis=1)
y_train_new = train_df_cleaned[TARGET]
X_test_new = test_df_cleaned.drop([TARGET, 'ID'], axis=1)
y_test_new = test_df_cleaned[TARGET]

print(f"Shape of new X_train: {X_train_new.shape}")
print(f"Shape of new y_train: {y_train_new.shape}")
print(f"Shape of new X_test: {X_test_new.shape}")
print(f"Shape of new y_test: {y_test_new.shape}")

# Recalculate scale_pos_weight for the new training set
count_negative_samples_new = y_train_new.value_counts()[0]
count_positive_samples_new = y_train_new.value_counts()[1]
scale_pos_weight_new = count_negative_samples_new / count_positive_samples_new
print(f"Recalculated scale_pos_weight: {scale_pos_weight_new:.2f}")

# Initialize the XGBoost Classifier with scale_pos_weight and modest regularization
model_regularized = xgb.XGBClassifier(
    objective='binary:logistic',  # For binary classification
    eval_metric='logloss',        # Evaluation metric
    random_state=42,
    scale_pos_weight=scale_pos_weight_new, # Apply the calculated weight
    # Add regularization parameters
    learning_rate=0.05,           # Step size shrinkage to prevent overfitting
    n_estimators=500,             # Number of boosting rounds
    max_depth=5,                  # Maximum depth of a tree
    subsample=0.7,                # Subsample ratio of the training instance
    colsample_bytree=0.7,         # Subsample ratio of columns when constructing each tree
    gamma=0.1,                    # Minimum loss reduction required to make a further partition on a leaf node
    reg_alpha=0.005,              # L1 regularization term on weights
    reg_lambda=0.1                # L2 regularization term on weights
)

# Train the regularized model
model_regularized.fit(X_train_new, y_train_new)

print("XGBoost Model (with new features, balanced weights, and regularization) training complete.")

# Make predictions on the test set with the regularized model
y_pred_regularized = model_regularized.predict(X_test_new)

# Evaluate the regularized model
accuracy_regularized = accuracy_score(y_test_new, y_pred_regularized)
print(f"\nAccuracy (Regularized Model): {accuracy_regularized:.4f}")

print("\nClassification Report (Regularized Model):")
print(classification_report(y_test_new, y_pred_regularized))

print("\nConfusion Matrix (Regularized Model):")
display(pd.DataFrame(confusion_matrix(y_test_new, y_pred_regularized), index=['Actual No Default', 'Actual Default'], columns=['Predicted No Default', 'Predicted Default']))
```
